```python
import matplotlib.pyplot as plt
```

```python
##Sales by region
df.groupby("Region")["Sales"].sum().plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 16))

# 1. Sales by Category (Bar Chart)
df.groupby("Category")["Sales"].sum().plot(kind="bar", ax=axes[0], color="skyblue", edgecolor="black")
axes[0].set_title("Sales by Category", fontsize=14, fontweight="bold")
axes[0].set_xlabel("Category", fontsize=12)
axes[0].set_ylabel("Sales ($)", fontsize=12)
axes[0].tick_params(axis='x', rotation=0)

# 2. Profit by Category (Bar Chart)
df.groupby("Category")["Profit"].sum().plot(kind="bar", ax=axes[1], color="mediumseagreen", edgecolor="black")
axes[1].set_title("Profit by Category", fontsize=14, fontweight="bold")
axes[1].set_xlabel("Category", fontsize=12)
axes[1].set_ylabel("Profit ($)", fontsize=12)
axes[1].tick_params(axis='x', rotation=0)

# 3. Monthly Sales Trend (Line Chart)
# Converts Period index to string for clean plotting
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.index = monthly_sales.index.astype(str)

monthly_sales.plot(kind="line", ax=axes[2], marker="o", color="crimson", linewidth=2)
axes[2].set_title("Monthly Sales Trend", fontsize=14, fontweight="bold")
axes[2].set_xlabel("Month", fontsize=12)
axes[2].set_ylabel("Sales ($)", fontsize=12)
axes[2].tick_params(axis='x', rotation=45)
axes[2].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python

# Business Insights
# -The West region contributes the highest sales.
# -Some sub-categories generate losses despite good sales.
# -Sales are concentrated among a few top customers.
# -Discounts may be reducing profit in certain categories.
# -Monthly sales show seasonal peaks.
```
